```python
# Install missing packages in this Jupyter environment (only runs in notebook)
%pip install -q spacy

import spacy

# 1. Load Language Model
# Use 'en_core_web_sm' for a small, fast model, or 'en_core_web_md' for better accuracy.
try:
    nlp = spacy.load("en_core_web_sm")
except Exception:
    print("Downloading model...")
    spacy.cli.download("en_core_web_sm")
    nlp = spacy.load("en_core_web_sm")

# Sample Amazon Reviews (Use more realistic data if available, or generate a few)
reviews = [
    "I absolutely love this new Samsung Galaxy S23. The battery life is phenomenal!",
    "The generic 40-watt lightbulb was disappointing and burned out after one week.",
    "The build quality of the Acme brand vacuum cleaner is surprisingly cheap.",
    "Great customer service, but the X-Pro headphones arrived broken."
]

def analyze_review(text):
    doc = nlp(text)
    
    # --- A. Named Entity Recognition (NER) ---
    product_brands = []
    # Identify entities related to organizations (brands) or products.
    # Note: 'ORG' often catches product names like 'Samsung' or 'Acme'.
    for ent in doc.ents:
        if ent.label_ in ["ORG", "PRODUCT"]:
            product_brands.append((ent.text, ent.label_))
            
    # --- B. Rule-Based Sentiment Analysis ---
    # Define simple rules based on keyword lists (Positive, Negative)
    positive_words = ["love", "great", "excellent", "phenomenal", "surprisingly good", "happy"]
    negative_words = ["disappointing", "burned out", "cheap", "broken", "worst", "poor"]
    
    sentiment = "Neutral"
    text_lower = text.lower()
    
    # Simple counting logic
    pos_count = sum(word in text_lower for word in positive_words)
    neg_count = sum(word in text_lower for word in negative_words)

    if pos_count > neg_count:
        sentiment = "Positive"
    elif neg_count > pos_count:
        sentiment = "Negative"
        
    return product_brands, sentiment

# --- Output Analysis ---
print("--- SpaCy NLP Analysis ---")
for i, review in enumerate(reviews):
    brands, sentiment = analyze_review(review)
    print(f"\nReview {i+1}: {review}")
    print(f"  Extracted Entities: {brands}")
    print(f"  Rule-Based Sentiment: {sentiment}")

# Deliverable check: Include a screenshot of this output showing extracted entities and sentiment in your PDF report.
```
